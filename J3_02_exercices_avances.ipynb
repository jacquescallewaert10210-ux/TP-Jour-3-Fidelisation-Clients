```python
import pandas as pd
import numpy as np

# --- Chargement des 3 tables (nettoyées au Jour 1)
commandes = pd.read_csv("data/commandes.csv", parse_dates=["date_commande"])
clients = pd.read_csv("data/clients.csv", parse_dates=["date_inscription"])
produits = pd.read_csv("data/produits.csv")

# --- Jointures : une ligne = une ligne de commande enrichie
df = (commandes
      .merge(produits, on="produit_id", how="left")
      .merge(clients, on="client_id", how="left"))

# --- Indicateurs de base
df["montant"] = df["quantite"] * df["prix_unitaire"] * (1 - df["remise"])
df["cout"] = df["quantite"] * df["cout_unitaire"]
df["marge"] = df["montant"] - df["cout"]

# --- Règle métier : le CA ne compte que les commandes livrées
ventes = df[df["statut"] == "Livrée"].copy()
print(f"{len(df)} lignes au total, dont {len(ventes)} commandes livrées")
```
